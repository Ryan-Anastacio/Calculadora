```python
nome = str(input("Digite seu nome: "))
print(f"É um Prazer te ver aqui {nome}")
print("Está é a calculadora 3mil plus!")

# Começo do loop
while True:
    numero1 = int(input("Digite o primeiro número: "))
    numero2 = int(input("Digite o segundo número: "))

    #Calculos da minha calculadora
    calc_soma = (numero1 + numero2)
    calc_sub = (numero1 - numero2)
    calc_mult = (numero1 * numero2)
    if numero2 == 0:
        print("Não é possível dividir por zero!")
        calc_divi = float('inf')
    else:
        calc_divi = (numero1 / numero2)

    #lista de comando
    lista_comando = (calc_soma, calc_sub, calc_mult, calc_divi)

    print("\n--- Escolha uma Operação ---")
    print("1 - Soma")
    print("2 - Subtração")
    print("3 - Multiplicação")
    print("4 - Divisão")
    print("0 - Sair")

    #Mostrar e selecionar opções
    mostrar_ope = str(input("Digite sua opção: "))

    if mostrar_ope == "1":
        print(f"O resultado da soma é: {calc_soma}")
    elif mostrar_ope == "2":
        print(f"O resultado da subtração é: {calc_sub}")
    elif mostrar_ope == "3":
        print(f"O resultado da multiplicação é: {calc_mult}")
    elif mostrar_ope == "4":
        if numero2 == 0:
            print("Não é possível exibir o resultado da divisão, pois o segundo número é zero.")
        else:
            print(f"O resultado da divisão é: {calc_divi}")
    elif mostrar_ope == "0":
        print("Obrigado por usar a calculadora 3mil plus!")
        break # Sai do loop
    else:
        print("Opção Inválida. Por favor, escolha novamente.")
```
